```python
import ast

def read_syscall_file(file_path):
    """
    Reads one ADFA-LD syscall 3-tuple frequency file.
    Returns a dictionary:
    {
        ('3', '3', '3'): 24381,
        ('4', '4', '4'): 19218,
        ...
    }
    """

    features = {}

    with open(file_path, "r") as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            try:
                syscall_tuple, frequency = ast.literal_eval(line)

                syscall_tuple = tuple(syscall_tuple)
                frequency = int(frequency)

                features[syscall_tuple] = frequency

            except Exception:
                continue

    return features


# Test on first file
test_features = read_syscall_file(first_file)

print("Number of 3-tuples:", len(test_features))

print("\nFirst 10 features:")

for key, value in list(test_features.items())[:10]:
    print(key, ":", value)
```

```python
def get_files(folder):
    return [f for f in folder.rglob("*") if f.is_file()]

train_files = get_files(TRAIN_PATH)
validate_files = get_files(VALIDATE_PATH)
attack_files = get_files(ATTACK_PATH)

print("Normal training files:", len(train_files))
print("Normal validation files:", len(validate_files))
print("Attack files:", len(attack_files))
```

```python
def read_syscall_file(file_path):
    with open(file_path, "r") as f:
        content = f.read().strip()

    if not content:
        return []

    return [int(x) for x in content.split()]
```

```python
sequence = read_syscall_file(train_files[0])

print("First file:", train_files[0].name)
print("Number of syscalls:", len(sequence))
print("First 30 syscalls:")
print(sequence[:30])
```

```python
def extract_features(sequence):
    
    features = {}

    # -----------------------------
    # 1. Sequence length
    # -----------------------------
    features["sequence_length"] = len(sequence)

    # -----------------------------
    # 2. Syscall frequency
    # -----------------------------
    syscall_counts = Counter(sequence)

    for syscall, count in syscall_counts.items():
        features[f"syscall_{syscall}"] = count

    # -----------------------------
    # 3. Transition / bigram frequency
    # -----------------------------
    for i in range(len(sequence) - 1):
        bigram = (sequence[i], sequence[i + 1])
        key = f"bigram_{bigram[0]}_{bigram[1]}"
        
        features[key] = features.get(key, 0) + 1

    # -----------------------------
    # 4. 3-gram frequency
    # -----------------------------
    for i in range(len(sequence) - 2):
        trigram = (sequence[i], sequence[i + 1], sequence[i + 2])
        key = f"trigram_{trigram[0]}_{trigram[1]}_{trigram[2]}"
        
        features[key] = features.get(key, 0) + 1

    return features
```

```python
test_features = extract_features(sequence)

print("Number of features:", len(test_features))

print("\nFirst 20 features:")

for key, value in list(test_features.items())[:20]:
    print(key, ":", value)
```

```python
normal_training_data = []

for file_path in train_files:
    
    sequence = read_syscall_file(file_path)
    
    features = extract_features(sequence)
    
    normal_training_data.append(features)

print("Files processed:", len(normal_training_data))
```

```python
X_train = pd.DataFrame(normal_training_data)

print("Shape before filling missing values:")
print(X_train.shape)

X_train = X_train.fillna(0)

print("\nShape after preprocessing:")
print(X_train.shape)

display(X_train.head())
```

```python
model = IsolationForest(
    n_estimators=200,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

model.fit(X_train)

print("Isolation Forest training completed!")
```

```python
normal_validation_data = []

for file_path in validate_files:
    
    sequence = read_syscall_file(file_path)
    
    features = extract_features(sequence)
    
    normal_validation_data.append(features)

X_normal_test = pd.DataFrame(normal_validation_data)

X_normal_test = X_normal_test.reindex(
    columns=X_train.columns,
    fill_value=0
)

print("Normal validation shape:", X_normal_test.shape)
```

```python
attack_data = []

for file_path in attack_files:
    
    sequence = read_syscall_file(file_path)
    
    features = extract_features(sequence)
    
    attack_data.append(features)

X_attack = pd.DataFrame(attack_data)

X_attack = X_attack.reindex(
    columns=X_train.columns,
    fill_value=0
)

print("Attack data shape:", X_attack.shape)
```

```python
normal_predictions = model.predict(X_normal_test)

attack_predictions = model.predict(X_attack)

print("Normal predictions:")
print(Counter(normal_predictions))

print("\nAttack predictions:")
print(Counter(attack_predictions))
```

```python
y_true = np.concatenate([
    np.zeros(len(X_normal_test)),
    np.ones(len(X_attack))
])

y_pred = np.concatenate([
    (normal_predictions == -1).astype(int),
    (attack_predictions == -1).astype(int)
])

print("0 = Normal")
print("1 = Attack")

print("\nTrue labels:", Counter(y_true))
print("Predicted labels:", Counter(y_pred))
```

```python
print("Accuracy:")
print(accuracy_score(y_true, y_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_true,
        y_pred,
        target_names=["Normal", "Attack"]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_true, y_pred))
```

```python
# Get anomaly scores

normal_scores = model.decision_function(X_normal_test)
attack_scores = model.decision_function(X_attack)

print("NORMAL scores:")
print("Minimum:", normal_scores.min())
print("Maximum:", normal_scores.max())
print("Mean:", normal_scores.mean())

print("\nATTACK scores:")
print("Minimum:", attack_scores.min())
print("Maximum:", attack_scores.max())
print("Mean:", attack_scores.mean())
```

```python
# Combine normal and attack scores

all_scores = np.concatenate([
    normal_scores,
    attack_scores
])

all_true = np.concatenate([
    np.zeros(len(normal_scores)),   # Normal = 0
    np.ones(len(attack_scores))     # Attack = 1
])

# Try many thresholds
thresholds = np.linspace(
    all_scores.min(),
    all_scores.max(),
    200
)

best_threshold = None
best_f1 = 0

from sklearn.metrics import f1_score

for threshold in thresholds:

    predictions = (all_scores < threshold).astype(int)

    f1 = f1_score(all_true, predictions)

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold

print("Best threshold:", best_threshold)
print("Best F1-score:", best_f1)
```

```python
# Apply the tuned threshold

normal_predictions = (
    normal_scores < best_threshold
).astype(int)

attack_predictions = (
    attack_scores < best_threshold
).astype(int)

# Combine predictions
y_true = np.concatenate([
    np.zeros(len(normal_predictions)),
    np.ones(len(attack_predictions))
])

y_pred = np.concatenate([
    normal_predictions,
    attack_predictions
])

print("Classification Report:\n")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=["Normal", "Attack"],
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_true, y_pred))
```
